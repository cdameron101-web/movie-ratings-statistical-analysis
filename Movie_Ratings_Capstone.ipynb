```python
import pandas as pd

print("Movie Ratings Capstone Project")
print("Pandas is ready!")
```

```python
movies = {
    "Movie": [
        "Scarface",
        "The Matrix",
        "Inception",
        "Goodfellas",
        "Scarface"
    ],
    "Rating": [9, 8, None, 10, 9]
}

df = pd.DataFrame(movies)
print(df)
```

```python
# Check for missing values
print("Missing Values:")
print(df.isnull().sum())

# Check for duplicates
print("\nDuplicate Rows:")
print(df.duplicated().sum())

# Clean the dataset
df_clean = df.dropna().drop_duplicates()


# Display cleaned data
print("\nCleaned Movie Dataseet:")
print(df_clean)
```

```python
from numpy import average
# Calculate movie rating statistics

average_rating = df_clean["Rating"].mean()
lowest_rating = df_clean["Rating"].min()
highest_rating = df_clean["Rating"].max()

print("Average Rating:", average_rating)
print("Lowest Movie Rating:", lowest_rating)
print("Highest Movie Rating:", highest_rating)
```

```python

# Count how often each rating appears
rating_counts = df_clean["Rating"].value_counts()

print("Movie Rating Distribution:")
print(rating_counts)

# Generate additional statistics
print("\nStatistical Summary:")
print(df_clean["Rating"].describe())
```

```python

import matplotlib.pyplot as plt

# Create a bar chart
plt.bar(df_clean["Movie"], df_clean["Rating"])

# Add labels and title
plt.title("Movie Ratings Analysis")
plt.xlabel("Movie")
plt.ylabel("Rating")

# Set rating scale
plt.ylim(0, 10)

# Display the chart
plt.show()
```

```python

import numpy as np

# Make results reproducible
np.random.seed(42)

# Simulate 100 movie ratings
ratings = np.random.normal(
    loc=8.2,
    scale=1.5,
    size=100
)

# Keep ratings between 1 and 10
ratings = np.clip(ratings, 1, 10)

print("First 10 Movie Ratings:")
print(ratings[:10])

print("\nTotal Ratings:", len(ratings))
```

```python

# Calculate the sample mean
sample_mean = np.mean(ratings)

# Calculate sample standard deviation
sample_std = np.std(ratings, ddof=1)

# Calculate sample size
sample_size = len(ratings)

# Calculate standard error
standard_error = sample_std / np.sqrt(sample_size)

print("Sample Mean:", round(sample_mean, 2))
print("Sample Standard Deviation:", round(sample_std, 2))
print("Sample Size:", sample_size)
print("Standard Error:", round(standard_error, 2))
```

```python

# Set the 95% confidence level
z_critical = 1.96

# Calculate margin of error
margin_of_error = z_critical * standard_error

# Calculate confidence interval
lower_bound = sample_mean - margin_of_error
upper_bound = sample_mean + margin_of_error

# Display results
print("Sample Mean:", round(sample_mean, 2))
print("Margin of Error:", round(margin_of_error, 2))

print("\n95% Confidence Interval:")
print(round(lower_bound, 2), "to", round(upper_bound, 2))
```

```python

# Generate our statistical summary

print("MOVIE RATINGS STATISTICAL REPORT")
print("--------------------------------")

print(f"Sample Size: {sample_size}")
print(f"Average Rating: {sample_mean:.2f}")
print(f"Standard Deviation: {sample_std:.2f}")
print(f"Standard Error: {standard_error:.2f}")

print(
    f"95% Confidence Interval: "
    f"{lower_bound:.2f} to {upper_bound:.2f}"
)
```

```python

# Generate a statistical conclusion

print("MOVIE RATINGS CONCLUSION")
print("------------------------")

print(
    f"Our sample of {sample_size} movie ratings "
    f"had an average rating of {sample_mean:.2f}."
)

print(
    f"The estimated 95% confidence interval "
    f"ranges from {lower_bound:.2f} "
    f"to {upper_bound:.2f}."
)

print(
    "This interval provides a range of plausible "
    "values for the population's average rating."
)
```

```python

# Final Movie Ratings Report

print("=" * 45)
print("MOVIE RATINGS - FINAL ANALYSIS REPORT")
print("=" * 45)

print(f"Total Ratings Analyzed: {sample_size}")
print(f"Average Rating: {sample_mean:.2f}")
print(f"Standard Deviation: {sample_std:.2f}")
print(f"Standard Error: {standard_error:.2f}")

print("\n95% CONFIDENCE INTERVAL")
print(f"Lower Bound: {lower_bound:.2f}")
print(f"Upper Bound: {upper_bound:.2f}")

print("\nFINAL CONCLUSION")

if sample_mean >= 7:
    print("The sample has a relatively high average rating.")
else:
    print("The sample has a relatively low average rating.")

print(
    "The confidence interval estimates the "
    "population's average rating, assuming "
    "appropriate sampling conditions."
)

print("\nAnalysis Complete!")
```
